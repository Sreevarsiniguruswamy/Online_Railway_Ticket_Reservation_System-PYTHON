```python
class Cancellation(Book_Ticket,Payment):
    def __init__(self, ticket_id):
        self.ticket_id = ticket_id
        self.is_cancelled = False
        self.cancellation_reason = ""
        self.refund_amount = 0

    def cancel_ticket(self, reason):
        if not self.is_cancelled:
            print(f"Ticket {self.ticket_id} has been cancelled successfully.")
            self.is_cancelled = True
            self.cancellation_reason = reason
            self.amount = self.calculate_refund()
        else:
            print(f"Ticket {self.ticket_id} is already cancelled.")

    def calculate_refund(self):
        # Implement refund calculation logic
        amount=int(input("enter your paid amount: "))
        print("your refund amount of ",amount,"will be refunded within 2 days")

    def __str__(self):
        return f"Ticket {self.ticket_id} - Cancelled: {self.is_cancelled}, Refund: {self.refund_amount}"
```

```python
import random
import string
from datetime import datetime, timedelta

class UserAlreadyExistsError(Exception):
    pass

class UserNotFoundError(Exception):
    pass

class AadhaarNameMismatchError(Exception):
    pass


class User:
    user_list = []
    booking_history = []

    def __init__(self, Firstname, Lastname, Gender, Date_of_Birth, Mobile_number, Email_id, Country, National_identity_card_number, Captcha,Aadhaar_card_name):
        self.Firstname = Firstname
        self.Lastname = Lastname
        self.Gender = Gender
        self.Date_of_Birth = Date_of_Birth
        self.Mobile_number = Mobile_number
        self.Email_id = Email_id
        self.Country = Country
        self.National_identity_card_number = National_identity_card_number
        self.Captcha = Captcha
        self.Aadhaar_name = Aadhaar_card_name

    def display_info(self):
        print(f"Name: {self.Firstname} {self.Lastname}")
        print(f"Gender: {self.Gender}")
        print(f"DOB: {self.Date_of_Birth}")
        print(f"Mobile Number: {self.Mobile_number}")
        print(f"Email Address: {self.Email_id}")
        print(f"Country: {self.Country}")
        print(f"National Identity Card Number: {self.National_identity_card_number}")
        print(f"Name in Aadhar card: {self.Aadhaar_card_name}")
    @classmethod
    def generate_captcha(cls):
        return ''.join(random.choices(string.ascii_letters + string.digits, k=6))

    @classmethod
    def register_user(cls, firstname, lastname, gender, dob, mobile, email, country, id_card,Aadhaar_card_name):
         # Aadhaar name validation: Check if Aadhaar Name matches first and last name
        full_name = f"{firstname} {lastname}"
        if Aadhaar_card_name != full_name:
            raise AadhaarNameMismatchError(f"Name in Aadhar card '{Aadhaar_card_name}' does not match the provided name '{full_name}'.")

        captcha = cls.generate_captcha()
        print(f"Captcha for registration: {captcha}")
        user_input_captcha = input("Please enter the CAPTCHA: ")
        
        if user_input_captcha != captcha:
            print("Captcha validation failed!")
            return None
        # Date of Birth validation
        try:
            dob_obj = datetime.strptime(dob, "%d/%m/%Y")
        except ValueError:
            print("Invalid date format. Please enter in DD/MM/YYYY format.")
            return None    

        for user in cls.user_list:
            if user.Email_id == email:
                raise UserAlreadyExistsError(f"User with email '{email}' is already registered.")
        
        new_user = User(firstname, lastname, gender, dob, mobile, email, country, id_card, captcha,Aadhaar_card_name)
        cls.user_list.append(new_user)
        print(f"User {firstname} {lastname} successfully registered!")
        return new_user

    @classmethod
    def login(cls, email):
        captcha = cls.generate_captcha()
        print(f"Captcha for login: {captcha}")
        user_input_captcha = input("Please enter the CAPTCHA: ")

        if user_input_captcha != captcha:
            print("Captcha validation failed!")
            return None

        for user in cls.user_list:
            if user.Email_id == email:
                print(f"Login successful for {user.Firstname} {user.Lastname}.")
                return user
        raise UserNotFoundError("User not found or incorrect captcha.")

    @classmethod
    def display_all_users(cls):
        print("\n--- Registered Users ---")
        for user in cls.user_list:
            user.display_info()

class BookTicket:
    def __init__(self, user, from_place, to_place, members, coach_no, dep_date, arr_date, train_class, category, food_choice=None):
        self.user = user
        self.from_place = from_place
        self.to_place = to_place
        self.members = members
        self.coach_no = coach_no
        self.dep_date = dep_date
        self.arr_date = arr_date
        self.train_class = train_class
        self.category = category
        self.food_choice = food_choice
        self.payment_deadline = datetime.now() + timedelta(hours=24)  # Payment deadline 24 hours from booking
        self.is_paid = False  # Initially, payment is not made
        
        User.booking_history.append(self)

    def calculate_payment(self, journey_time_hours):
        # Dummy payment logic: You can change this as per your pricing model
        base_fare = 500
        class_multiplier = {"1A": 3, "2A": 2, "SL": 1, "FC": 2.5}
        food_cost = 200 if self.food_choice.lower() == "yes" else 0
        total_cost = (base_fare * class_multiplier.get(self.train_class, 1)) * self.members + food_cost
        return total_cost
    def display_booking(self):
        print(f"\n--- Booking for {self.user.Firstname} {self.user.Lastname} ---")
        print(f"From: {self.from_place}, To: {self.to_place}, Coach: {self.coach_no}")
        print(f"Departure: {self.dep_date}, Arrival: {self.arr_date}, Class: {self.train_class}, Category: {self.category}")
        print(f"Food Choice: {self.food_choice if self.food_choice else 'No food booked'}")
        print(f"Payment Deadline: {self.payment_deadline.strftime('%Y-%m-%d %H:%M:%S')}")
        print(f"Payment Status: {'Completed' if self.is_paid else 'Pending'}")

    @classmethod
    def view_booking_history(cls, user):
        if not User.booking_history:
            print("No booking history available.")
            return
        print(f"\n--- Booking History for {user.Firstname} {user.Lastname} ---")
        for booking in User.booking_history:
            if booking.user == user:
                booking.display_booking()

    @classmethod
    def cancel_booking(cls, user):
        for booking in reversed(User.booking_history):
            if booking.user == user:
                print(f"\nCancelling booking from {booking.from_place} to {booking.to_place}.")
                User.booking_history.remove(booking)
                return
        print("No bookings to cancel.")

class Train:
    Train_list = []

    def __init__(self, train_code, total_seats, train_name, start_time, end_time,seats):
        self.train_code = train_code
        self.total_seats = total_seats
        self.train_name = train_name
        self.start_time = start_time
        self.end_time = end_time
        self.seats = {
            "Upper": total_seats // 3,
            "Lower": total_seats // 3,
            "Middle": total_seats // 3
        }
        self.food_menu = {
            'veg': ['Veg Thali', 'Paneer Biryani', 'Dal Fry with Roti'],
            'non_veg': ['Chicken Biryani', 'Egg Curry with Rice', 'Fish Fry with Chapati'],
            'Snacks': ['Bhel puri', 'Burger', 'Chilli Gobi', 'Mushroom Fry']
        }
        Train.Train_list.append(self)

    def display_train_details(self):
        print(f"\nTrain Code: {self.train_code}")
        print(f"Train Name: {self.train_name}")
        print(f"Start Time: {self.start_time}")
        print(f"End Time: {self.end_time}")
        print(f"Total Seats: {self.total_seats}")
        print(f"Seat Availability: {self.seats}")

    @classmethod
    def display_all_trains(cls):
        for train in cls.Train_list:
            train.display_train_details()

    def check_availability(self):
        available = sum(self.seats.values())
        if available > 0:
            print(f"\nTotal Tickets available: {available}.")
            print(f"Upper Berth: {self.seats['Upper']} seats left.")
            print(f"Lower Berth: {self.seats['Lower']} seats left.")
            print(f"Middle Berth: {self.seats['Middle']} seats left.")
            return True
        else:
            print("\nNo tickets available.")
            return False

    def check_category_availability(self, category, members):
        if category in self.seats:
            if self.seats[category] >= members:
                return True
            else:
                print(f"\nOnly {self.seats[category]} seats left in {category} Berth.")
                return False
        else:
            print(f"\nInvalid category: {category}.")
            return False

    def update_seat_availability(self, category, members):
        if self.check_category_availability(category, members):
            self.seats[category] -= members
            print(f"\n{members} seat(s) booked in {category} Berth. Remaining {category} seats: {self.seats[category]}")
            return True
        else:
            print(f"\nBooking failed due to insufficient seats in {category} Berth.")
            return False

    def display_food_menu(self):
        
        veg_options = ', '.join(self.food_menu['veg'])
        non_veg_options = ', '.join(self.food_menu['non_veg'])
        snacks_options = ', '.join(self.food_menu['Snacks'])
        print(f"Available Veg Options: {veg_options}")
        print(f"Available Non-Veg Options: {non_veg_options}")
        print(f"Available Snacks: {snacks_options}")



class Payment:
    def __init__(self, pay_mode, amount, pay_date):
        self.pay_mode = pay_mode
        self.amount = amount
        self.pay_date = pay_date

    def display_payment(self):
        print(f"\n--- Payment Details ---")
        print(f"Payment Mode: {self.pay_mode}")
        print(f"Amount Paid: {self.amount}")
        print(f"Payment Date: {self.pay_date}")

class Notification:
    @staticmethod
    def send_notification(user, message):
        print(f"Notification to {user.Firstname}: {message}")

    @staticmethod
    def check_notifications(user):
        today = datetime.now()
        birthday_month = int(user.Date_of_Birth.split("/")[1])
        birthday_day = int(user.Date_of_Birth.split("/")[0])

        if today.day == birthday_day and today.month == birthday_month:
            Notification.send_notification(user, "Happy Birthday! Enjoy special offers from IRCTC!")

        if today.month in [12, 10]:  # Example festive months
            Notification.send_notification(user, "Check out our special offers during the festive season!")

# Main Function for Register/Login loop with additional features
def main():
    print("Welcome to IRCTC!!!")
    Train_list = [
        Train("12345", 500, "Rajdhani Express", "09:00 AM", "05:00 PM", 150),
        Train("22334", 400, "Kovai Express", "07:00 AM", "03:00 PM", 120),
        Train("33456", 600, "Chennai Express", "10:00 AM", "06:00 PM", 200),
        Train("44567", 350, "Nilgiri Express", "06:00 AM", "02:00 PM", 100)
    ]
     # Create a list of coach numbers from A1 to Z50
    available_coaches = [f"{letter}{number}" for letter in 'ABCDEFGHIJKLMNOPQRSTUVWXYZ' for number in range(1, 51)]

    while True:
        print("\n1. New User Registration\n2. Login\n3. Admin Login\n4. Exit")
        choice = input("Enter your choice (1-4): ")

        if choice == '1':
            try:
                fname = input("First Name: ")
                lname = input("Last Name: ")
                gender = input("Gender: ")
                 # Add DOB validation here
                dob = input("Date of Birth (DD/MM/YYYY): ")
                try:
                    dob_obj = datetime.strptime(dob, "%d/%m/%Y")
                except ValueError:
                    print("Invalid date format. Please enter in DD/MM/YYYY format.")
                    continue
                    
                mobile = input("Mobile Number: ")
                email = input("Email ID: ")
                country = input("Country: ")
                id_card = input("National Identity Card Number (AADHAR, PAN, Voter ID): ")
                Aadhaar_card_name = input("Aadhaar Name (as per Aadhaar): ")

                # Attempt to register user, Aadhaar name validation included
                user = User.register_user(fname, lname, gender, dob, mobile, email, country, id_card, Aadhaar_card_name)
            except UserAlreadyExistsError as e:
                print(e)
            except AadhaarNameMismatchError as e:
                print(e)

        elif choice == '2':
            try:
                email = input("Enter your email: ")
                user = User.login(email)

                if not user:
                    continue  # Failed login

                # Proceed after login
                while True:
                    print("\n--- User Menu ---")
                    print("1. Book a Ticket")
                    print("2. View Booking History")
                    print("3. Cancel Last Booking")
                    print("4. Check Train Availability")
                    print("5. Check Notifications")
                    print("6. Logout")

                    user_choice = input("Enter your choice (1-6): ")

                    if user_choice == '1':
                        # Choose a train to book
                        print("\n--- Available Trains ---")
                        Train.display_all_trains()
    
                        selected_train_code = input("Enter the Train Code you want to book: ")
                        train = next((t for t in Train.Train_list if t.train_code == selected_train_code), None)

                        if not train:
                            print("Invalid train code. Please try again.")
                            continue
                        
                        if train.check_availability():
                            from_place = input("From: ")
                            to_place = input("To: ")
                            try:
                                members = int(input("Number of members: "))
                                if members <= 0:
                                    print("Number of members must be at least 1.")
                                    continue
                            except ValueError:
                                print("Please enter a valid number for members.")
                                continue
                            # Check if requested members exceed available seats
                            # Sum up total available seats from all categories (Upper, Lower, Middle)
                            total_available_seats = sum(train.seats.values())

    # Check if requested members exceed available seats
                            if members > total_available_seats:
                                print(f"Only {total_available_seats} seats are available. Please reduce the number of members.")
                                continue  # Skip the rest of the booking process and go back to the menu
                            
                            available_coaches = [f"{letter}{number}" for letter in 'ABCDEFGHIJKLMNOPQRSTUVWXYZ' for number in range(1, 51)]
                            
                            # Randomly assign a coach number
                            # Assign a coach number
                            if not available_coaches:
                                print("\nNo coaches available.")
                                continue

                            random_coach = random.choice(available_coaches)
                            available_coaches.remove(random_coach)  # Remove to prevent double booking
                            print(f"Assigned coach number: {random_coach}")
                            
                            dep_date = input("Date of departure (DD/MM/YYYY): ")
                            arr_date = input("Date of arrival (DD/MM/YYYY): ")
                            train_class = input("Train class (AC FIRST CLASS(1A),AC TIER(2A),SLEEPER CLASS(SL),FIRST CLASS(FC): ")
                            category = input("Category (Lower Berth, Upper Berth, Middle Berth, Tatkal, Person with disability): ")

# Validate category input
                            valid_categories = ["Upper", "Lower", "Middle", "Tatkal", "Person with disability"]
                            if category not in valid_categories:
                                print("\nInvalid category selected.")
                                available_coaches.append(random_coach)  # Re-add coach as booking didn't proceed
                                continue

# Handle special categories (Tatkal and Person with disability)
                            if category in ["Tatkal", "Person with disability"]:
    # Assign category without checking seat availability since it's not tracked
                                print(f"\nYou have selected {category}. Proceeding with booking...")
    # Additional logic for handling Tatkal or Person with disability booking can go here
                            else:
    # Check availability in the selected category (Upper, Lower, Middle)
                                if not train.check_category_availability(category, members):
                                    available_coaches.append(random_coach)  # Re-add coach as booking didn't proceed
                                    continue  # Go back to user menu
                                else:
        # Proceed with booking process if seats are available in the selected category
                                    train.update_seat_availability(category, members)
                                    print(f"\nBooking successful for {members} seat(s) in {category}.")

                            food_choice = input("Food booking needed? (Yes/No): ")
                            # Ask the user if they want to book food


# If the user selects "Yes", display the food menu
                        if food_choice == "Yes":
                            print("\nDisplaying Food Menu:")
                            train.display_food_menu()  # Assuming 'train' is the object of Train class being used in the booking
                            veg_or_nonveg = input("\nWould you like to choose Veg, Non-Veg, or Snacks? ").strip().title()

                            if veg_or_nonveg == "Veg":
                                print(f"Available Veg Options: {', '.join(train.food_menu['veg'])}")
                                selected_food = input("\nPlease select a veg item: ").strip().title()
                                if selected_food not in train.food_menu['veg']:
                                    print("Invalid choice. Please try again.")
                                    continue
                            elif veg_or_nonveg == "Non-Veg":
                                print(f"Available Non-Veg Options: {', '.join(train.food_menu['non_veg'])}")
                                selected_food = input("\nPlease select a non-veg item: ").strip().title()
                                if selected_food not in train.food_menu['non_veg']:
                                    print("Invalid choice. Please try again.")
                                    continue
                            elif veg_or_nonveg == "Snacks":
                                print(f"Available Snacks: {', '.join(train.food_menu['Snacks'])}")
                                selected_food = input("\nPlease select a snack: ").strip().title()
                                if selected_food not in train.food_menu['Snacks']:
                                    print("Invalid choice. Please try again.")
                                    continue
                            else:
                                print("Invalid category selected. Please try again.")
                        else:
                            print("Food booking skipped.")

                                
                            
                            # Simulate journey time in hours
                            journey_time_hours = random.randint(1, 24)
                            
                            # Calculate payment
                            ticket = BookTicket(user, from_place, to_place, members, random_coach, dep_date, arr_date, train_class, category, food_choice)
                        
                            payment_amount = ticket.calculate_payment(journey_time_hours)
                            print(f"Total payment for {members} members is: Rs. {payment_amount}")
                            # Display payment deadline
                            print(f"Please complete the payment by: {ticket.payment_deadline.strftime('%Y-%m-%d %H:%M:%S')}")

    # Check if the user is making the payment before the deadline
                            current_time = datetime.now()

                            if current_time > ticket.payment_deadline:
                                print("Payment deadline has passed. Booking is cancelled.")
        # Remove the booking from booking history
                                User.booking_history.remove(ticket)
                                return
                                
                            else:
                               # Process payment
                                payment_mode = input("\nEnter payment mode (Credit Card, Debit Card, UPI, G-PAY): ").strip().title()
                                if payment_mode not in ["Credit Card", "Debit Card", "Upi", "G-pay"]:
                                    print("\nInvalid payment mode selected.")
                                    User.booking_history.remove(ticket)
                                    available_coaches.append(random_coach)  # Re-add coach as booking didn't proceed
                                    continue

                                payment = Payment(payment_mode, payment_amount, datetime.now())
                                payment.display_payment()

                                # Mark the ticket as paid
                                ticket.is_paid = True
                                
                                # Update seat availability in the selected category
                                if category in ["Upper ", "Lower", "Middle"]:
                                    if not train.update_seat_availability(category, members):
                                        print("\nBooking failed due to insufficient seats.")
                                        User.booking_history.remove(ticket)
                                        available_coaches.append(random_coach)  # Re-add coach as booking didn't proceed
                                        continue

                                print(f"\n{members} ticket(s) booked successfully! Remaining seats in {category} Berth: {train.seats.get(category, 'N/A')}")

                                # Optionally, send a notification
                                Notification.send_notification(user, "Your payment is successful!!! Enjoy your journey!!!")
                    

                    elif user_choice == '2':
                        BookTicket.view_booking_history(user)

                    elif user_choice == '3':
                        BookTicket.cancel_booking(user)

                    elif user_choice == '4':
                        train.check_availability()

                    elif user_choice == '5':
                        Notification.check_notifications(user)

                    elif user_choice == '6':
                        print("Logging out...")
                        break
            except UserNotFoundError as e:
                print(e)
                    
        elif choice == '3':
            admin_password = "admin123"  # Simple admin password
            entered_password = input("Enter admin password: ")

            if entered_password == admin_password:
                print("\n--- Admin Menu ---")
                User.display_all_users()
            else:
                print("Incorrect password. Access denied.")

        elif choice == '4':
            print("Exiting the application.")
            break

        else:
            print("Invalid choice. Please select again.")

if __name__ == "__main__":
    main()

         
```
